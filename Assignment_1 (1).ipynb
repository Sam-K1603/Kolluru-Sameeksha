## Question 2: Lists and Tuples

```python
numbers = [12, 45, 7, 23, 89, 34, 56, 3, 67, 21]
print("Original list      :", numbers)

# Add an element
numbers.append(100)
print("After adding 100   :", numbers)

# Remove an element
numbers.remove(7)
print("After removing 7   :", numbers)

# Insert an element at a specified position
numbers.insert(2, 50)
print("After inserting 50 at index 2:", numbers)

# Maximum and minimum
print("Maximum element    :", max(numbers))
print("Minimum element    :", min(numbers))

# Convert list into tuple
numbers_tuple = tuple(numbers)
print("List as tuple      :", numbers_tuple)
print("Type               :", type(numbers_tuple))
```

## Question 3: Strings

```python
text = ask("Enter a string: ", "Madam")

# Number of characters
print("Number of characters:", len(text))

# Vowels and consonants
vowels = "aeiouAEIOU"
vowel_count = sum(1 for ch in text if ch in vowels)
consonant_count = sum(1 for ch in text if ch.isalpha() and ch not in vowels)
print("Vowels              :", vowel_count)
print("Consonants          :", consonant_count)

# Reverse
print("Reversed string     :", text[::-1])

# Palindrome check (ignores case, spaces and punctuation)
cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
if cleaned == cleaned[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

## Question 5: GPA Calculator
GPA = Σ(Grade Point × Credit) / Σ Credits

```python
n = int(ask("Enter number of courses: ", "3"))

total_points = 0
total_credits = 0

for i in range(1, n + 1):
    grade_point = float(ask(f"Course {i} - enter grade point: ", [9, 8, 10][(i - 1) % 3]))
    credit = float(ask(f"Course {i} - enter credit: ", [4, 3, 2][(i - 1) % 3]))
    total_points += grade_point * credit
    total_credits += credit

if total_credits > 0:
    gpa = total_points / total_credits
    print(f"\nTotal credits : {total_credits}")
    print(f"SGPA / GPA    : {gpa:.2f}")
else:
    print("Total credits must be greater than zero.")
```

## Question 6: Functions and Iteration

```python
def factorial(n):
    """Return the factorial of n."""
    if n < 0:
        return None
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


def fibonacci(n):
    """Return a list with the first n Fibonacci numbers."""
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series


def is_prime(n):
    """Return True if n is prime, otherwise False."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


num = int(ask("Enter a number for factorial: ", "5"))
fact = factorial(num)
if fact is None:
    print("Factorial is not defined for negative numbers.")
else:
    print(f"Factorial of {num} = {fact}")

terms = int(ask("Enter number of Fibonacci terms: ", "8"))
print("Fibonacci series:", fibonacci(terms))

p = int(ask("Enter a number to check for prime: ", "17"))
print(f"{p} is prime." if is_prime(p) else f"{p} is not prime.")
```

## Question 8: Exception Handling

```python
try:
    a = float(ask("Enter the numerator: ", "10"))
    b = float(ask("Enter the denominator: ", "0"))
    result = a / b
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Please enter valid numbers only.")
else:
    print(f"Result: {a} / {b} = {result}")
finally:
    print("Division program finished.")
```

## Question 10: Mini Programming Problem (Student Record Menu)

```python
records = {}
# Sample inputs used when INTERACTIVE = False
demo = iter(["1","Ravi","85","1","Anu","92","1","Ravi","2","Ravi","90","3","Anu","4","Ravi","5","x","6"])

while True:
    print("\n===== STUDENT RECORD MENU =====")
    print("1. Add a student")
    print("2. Update marks")
    print("3. Search for a student")
    print("4. Delete a student")
    print("5. Display all student records")
    print("6. Exit")

    choice = ask("Enter your choice (1-6): ", next(demo, "6")).strip()

    if choice == "1":
        name = ask("Enter student name: ", next(demo, "6")).strip()
        if name in records:
            print("Student already exists. Use option 2 to update marks.")
        else:
            try:
                records[name] = float(ask("Enter marks: ", next(demo, "6")))
                print("Student added.")
            except ValueError:
                print("Invalid marks. Please enter a number.")

    elif choice == "2":
        name = ask("Enter student name: ", next(demo, "6")).strip()
        if name in records:
            try:
                records[name] = float(ask("Enter new marks: ", next(demo, "6")))
                print("Marks updated.")
            except ValueError:
                print("Invalid marks. Please enter a number.")
        else:
            print("Student not found.")

    elif choice == "3":
        name = ask("Enter student name to search: ", next(demo, "6")).strip()
        if name in records:
            print(f"Found: {name} - Marks: {records[name]}")
        else:
            print("Student not found.")

    elif choice == "4":
        name = ask("Enter student name to delete: ", next(demo, "6")).strip()
        if name in records:
            del records[name]
            print("Student deleted.")
        else:
            print("Student not found.")

    elif choice == "5":
        if records:
            print(f"\n{'Name':<20}{'Marks':>8}")
            print("-" * 28)
            for name, marks in records.items():
                print(f"{name:<20}{marks:>8}")
        else:
            print("No records to display.")

    elif choice == "6":
        print("Exiting the program. Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 6.")
```
